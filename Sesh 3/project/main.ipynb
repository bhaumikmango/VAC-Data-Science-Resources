```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('student-mat-pass-or-fail.csv')
```

```python
# Display the first few rows and column information
print("--- DataFrame Head ---")
print(df.head())
print("\n--- DataFrame Info ---")
df.info()
```

```python
X = df.drop('pass', axis=1)
y = df['pass']
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=69)

print(f"\nTraining set size (X_train): {X_train.shape}")
print(f"Testing set size (X_test): {X_test.shape}")
```

```python
param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 2, 5, 10, 20, 50, 100],
    'solver': ['liblinear', 'lbfgs','newton-cg', 'newton-cholesky', 'sag', 'saga'], 
    'max_iter': [10, 100, 1000, 10000],
    }
```

```python
logreg = LogisticRegression(random_state=42)

grid_search = GridSearchCV(estimator=logreg, param_grid=param_grid, cv=8, scoring='accuracy', n_jobs=-1, verbose=1)

grid_search.fit(X_train, y_train)

best_params = grid_search.best_params_
best_score = grid_search.best_score_
best_estimator = grid_search.best_estimator_

print("\n--- GridSearchCV Results ---")
print(f"Best Hyperparameters found on the training set: {best_params}")
print(f"Best Cross-Validation Score: {best_score:.4f}")
best_estimator
```

```python
tuned_model = best_estimator

y_pred_tuned = tuned_model.predict(X_test)

accuracy_tuned = accuracy_score(y_test, y_pred_tuned)
conf_matrix_tuned = confusion_matrix(y_test, y_pred_tuned)
class_report_tuned = classification_report(y_test, y_pred_tuned)

print(f"\n--- Tuned Model Performance on Test Set ---")
print(f"Tuned Model Accuracy: {accuracy_tuned:.4f}")
print("\nClassification Report (Tuned Model):")
print(class_report_tuned)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix_tuned, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Fail (0)', 'Pass (1)'], yticklabels=['Fail (0)', 'Pass (1)'])
plt.title('Confusion Matrix (Tuned Model)')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.show()
```
